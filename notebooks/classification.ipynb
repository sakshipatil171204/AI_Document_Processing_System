```python
import pandas as pd
import os

ocr_folder = "../data/ocr_text"

data = []

for file in os.listdir(ocr_folder):
    file_path = os.path.join(ocr_folder, file)

    with open(file_path, "r", encoding="utf-8") as f:
        text = f.read()

    if "invoice" in file.lower():
        label = "Invoice"
    elif "marksheet" in file.lower():
        label = "Marksheet"
    elif "resume" in file.lower():
        label = "Resume"

    data.append({
        "filename": file,
        "text": text,
        "label": label
    })

df = pd.DataFrame(data)

df
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

df["clean_text"] = df["text"].apply(clean_text)

df[["filename", "label", "clean_text"]]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(df["clean_text"])
y = df["label"]

print("TF-IDF shape:", X.shape)
print("Labels:", y.tolist())
```

```python
features = vectorizer.get_feature_names_out()

print("Number of features:", len(features))
print("\nTF-IDF Features:")
print(features)
```

```python
tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=features,
    index=df["label"]
)

tfidf_df
```
